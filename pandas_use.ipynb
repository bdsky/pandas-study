```python
import pandas as pd

s1 = pd.Series([1,2,3,4], index = list('abcd'))
print(s1)
print(type(s1))
print('-'*30)

data = {
    'math': [22,33,44],
    'eng': [11,22,33],
    'chinese': [88,99,77]
}
s2 = pd.DataFrame(data)
print(s2)
print(type(s2))

print('-'*30)
print(s2['math'])
print(f'{type(s2['math']) = }')
print('-'*30)
print(s2[['math']])
print(f'{type(s2[['math']]) = }')

print('-'*30)
print(s2.columns)
print(s2.index)
print(s2.dtypes)
```
