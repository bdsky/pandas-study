## pandas基础
### pandas的数据结构
- pandas的两种基本数据结构： Series, DataFrame
- Series是带标签的一维数组
- DataFrame是二维表格，有行列标签，但是单看DataFrame的某一列就是Series类型（见下面s2['math]的类型输出）


```python
import pandas as pd

s1 = pd.Series([1,2,3,4], index = list('abcd'))
print(s1)
print(type(s1))
print('-'*30)

data = {
    'math': [22,33,44],
    'eng': [11,22,33],
    'chinese': [88,99,77]
}
s2 = pd.DataFrame(data)
print(s2)
print(type(s2))

print('-'*30)
print(s2['math'])
print(f'{type(s2['math']) = }')
print('-'*30)
print(s2[['math']])
print(f'{type(s2[['math']]) = }')

print('-'*30)
print(s2.columns)
print(s2.index)
print(s2.dtypes)
```

#### pandas的列选择
- DataFrame的列选择语法是df[xx]
- xx是一个普通的值时返回某一列，返回类型为Series
- xx是列表时，返回多个列，也就是一个新的DataFrame类型数据

```python
s3 = pd.DataFrame(data)
print('-'*40)
print(s3['math'], '\n')
print(f'{type(s3['math']) = }\t{s3['math'].shape = }', '\n')
print(s3[['math']], '\n')
print(f'{type(s3[['math']]) = }\t{s3[['math']].shape = }', '\n')
print(s3[['math', 'chinese']], '\n')
print(f'{type(s3[['math', 'chinese']]) = }\t{s3[['math', 'chinese']].shape}')
```

### pandas的（位置|标签）选择
- DataFrame的行列都有默认的位置编号，从0开始
- var.iloc[]使用**行列位置**编号来选择
- var.loc[]使用**行列标签**来选择，但是注意默认情况下行的位置标签和位置编号是一样的，从0开始
- 当使用set_index()方法将某列数据作用行标签使用时，此时var.loc[]不能使用0，1..等作参数，因为0，1..仅表示位置了
- 使用set_index()后可以使用reset_index()恢复行标签为0, 1,....
- var.iloc[xx,xx]和var.loc[xx,xx]都使用','分隔行列参数，可以仅使用行参数或列参数，省略部分表示全选
- iloc和loc都可以使用切片，但是通常loc的切片两个边界都包含在内
- 注意区分轴编号，轴标签， 轴的名字

```python
data2 = {
    'student':['zs', 'ls', 'ww'],
    'math': [22,33,44],
    'eng': [11,22,33],
    'chinese': [88,99,77]
}
s4 = pd.DataFrame(data2)
print('-'*40)
print(f'{s4 = }', '\n')
print(s4.iloc[0], '\n')
print(s4.iloc[0:2], '\n')
print(s4.iloc[0, 2], '\n')
print(s4.iloc[0:2, 0:2], '\n')
print('-'*40)
print(s4.loc[0], '\n')
print(s4.loc[0:2], '\n')
print(s4.loc[0,'eng'], '\n')
print(f'{s4.loc[0:2, 'student': 'chinese'] = }', '\n')
print(f'{s4.loc[0:2, 'math': 'chinese'] = }', '\n')
print('-'*40)
s5 = s4.set_index('student')
print(f'{s5 = }', '\n')
print(f'{s5.iloc[0:2, 0:2] = }', '\n')
print(f'{s5.loc['zs': 'ls', 'math':'eng'] = } ', '\n')
```

### pandas的条件筛选
- 设df为DataFrame类型数据
- df[xx] > yy , 当df[xx]是Series类型时，这个表达时返回的也是Series类型，其中各个元素值为True or False;
- df[zz], 当zz是bool值Series时，df[zz]是根据bool值决定是否保留对应行
- 如果zz不是bool值Series，则Series和列表相同，用于选择列
- 多条件筛选使用 & | ~ （且， 或者， 取反）

```python
s6 = pd.DataFrame(data2)
print('-'*40)
print(f'{s6 = }', '\n')
print(f'{s6['math'] > 30 = }', '\n')
print(f'{type(s4['math'] > 30)}', '\n')
print('-'*40)
print(s6[s6['math'] > 30], '\n')
l1 = pd.Series(['math', 'eng'])
print(f'{s6[l1] = }', '\n')

cond = (s6['math'] > 30) & (s6['eng'] > 30)
print(f'{s6[cond] = }, \n{type(s6[cond]) = }',  '\n')
```
